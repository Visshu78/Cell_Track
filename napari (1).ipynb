```python
from pathlib import Path

data_dir = Path("cell_tracking_data")
data_dir.mkdir(exist_ok=True)

print("Data directory:", data_dir.absolute())
```

```python
import pooch

print("Pooch version:", pooch.__version__)
```

```python
import pooch

st_url = "doi:10.5281/zenodo.15852284/masks_pred.npz"

st_path = pooch.retrieve(
    url=st_url,
    fname="masks_pred.npz",
    known_hash=None,
    path=data_dir
)

print("Mask file downloaded:")
print(st_path)
```

```python
import numpy as np

masks_npz = np.load(st_path)

print("Files inside masks_pred.npz:")
print(masks_npz.files)

masks = masks_npz["masks"]

print("Mask shape:", masks.shape)
print("Mask data type:", masks.dtype)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 8))

plt.imshow(masks[0])

plt.title("Segmentation Mask — Frame 1")
plt.axis("off")

plt.show()
```

```python
unique_labels = np.unique(masks[0])

print("Unique labels:", unique_labels)
print("Number of cells:", len(unique_labels) - 1)
```

```python
for t in range(masks.shape[0]):
    num_cells = len(np.unique(masks[t])) - 1
    print(f"Frame {t}: {num_cells} cells")
```

```python
tif_url = "doi:10.5281/zenodo.17643282/01(1).zip"

tif_files = pooch.retrieve(
    url=tif_url,
    fname="01(1).zip",
    known_hash=None,
    path=data_dir,
    processor=pooch.Unzip()
)

print("Number of extracted files:", len(tif_files))
```

```python
import tifffile
import numpy as np

imgs = sorted(tif_files)

images = np.stack([
    tifffile.imread(filename)
    for filename in imgs
])

print("Images shape:", images.shape)
print("Images dtype:", images.dtype)
```

```python
print("Images shape:", images.shape)
print("Masks shape :", masks.shape)

print("\nNumber of image frames:", images.shape[0])
print("Number of mask frames :", masks.shape[0])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 8))

plt.imshow(images[0], cmap="gray")

plt.title("Raw Microscopy Image — Frame 1")
plt.axis("off")

plt.show()
```

```python
plt.figure(figsize=(8, 8))

plt.imshow(images[0], cmap="gray")
plt.imshow(masks[0], alpha=0.4)

plt.title("Raw Image + Segmentation Mask")
plt.axis("off")

plt.show()
```

```python
graph, *_ = model.track(
    imgs=images,
    masks=masks,
    mode="greedy"
)

print("Tracking completed!")
```

```python
print("Graph type:")
print(type(graph))
print("Graph:")
print(graph)
print("Graph attributes:")
print(dir(graph))
```

```python
first_node = list(graph.nodes)[0]

print("First node ID:", first_node)
print("\nNode information:")
print(graph.nodes[first_node])
```

```python
for node in list(graph.nodes)[:10]:
    print(node, "->", graph.nodes[node])
```

```python
print("Edge information:")
for successor in graph.successors(0):
    print(
        f"Node 0 -> Node {successor}:",
        graph.edges[0, successor]
    )
```

```python
from trackastra.tracking import graph_to_ctc

tracks_df, tracked_masks = graph_to_ctc(
    graph=graph,
    masks_original=masks
)

print("Conversion completed!")
```

```python
print(type(tracks_df))
print(tracks_df.head())
print("Shape:", tracks_df.shape)
print("Columns:")
print(tracks_df.columns.tolist())
```

```python
print("Type:", type(tracked_masks))
print("Shape:", tracked_masks.shape)
print("Data type:", tracked_masks.dtype)
```

```python
plt.figure(figsize=(8, 8))

plt.imshow(tracked_masks[0])

plt.title("Tracked Mask — Frame 1")
plt.axis("off")

plt.show()
```

```python
for t in range(tracked_masks.shape[0]):
    labels = np.unique(tracked_masks[t])
    num_cells = len(labels) - 1
    print(f"Frame {t}: {num_cells} tracked cells")
```

```python
import napari

viewer = napari.Viewer()

viewer.add_image(
    images,
    name="Raw Microscopy",
    colormap="gray"
)

viewer.add_labels(
    masks,
    name="Original Segmentation"
)

viewer.add_labels(
    tracked_masks,
    name="Tracked Cells"
)

napari.run()
```
